```python
def check_gender_by_name(name):
  if name[-1]=="a":
    gender = "female"
  elif name[-1]=="o":
    gender = "male"
  else:
    gender = "unknown"
  return gender
```

```python
def ph(c):
  return -math.log10(c)
```

```python
ph(0.001)
```

```python
def ph_evaluator(c=0.001):
  ''' Ph decision maker'''
  ph_value = ph(c)

  if ph_value < 7:
    decision = 'kiselo'
  else:
    decision = 'lužnato'

  return decision

ph_evaluator()
```

```python
def test_if_else(a, b):
  '''
  a, b su argumenti
  return 'a > b' or 'a is not > b'
  '''
  if a > b:
    print(a, ' > ', b)
  else:
    print('%d is not > %d' %(a, b))
```

```python
def rectangle_area(a,b):
  '''dokumentacija'''
  A = a*b
  if a == b:
    print('kvadrat')
  else:
    print('pravokutnik')
  return A
```
